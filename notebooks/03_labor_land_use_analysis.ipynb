```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

# Raw GitHub URLs
url_labour = "https://raw.githubusercontent.com/oxedanda/avcad_final_project/main/data/processed/agricultural_labour_clean.csv"
url_area = "https://raw.githubusercontent.com/oxedanda/avcad_final_project/main/data/processed/agricultural_holdings_area_clean.csv"
url_crop = "https://raw.githubusercontent.com/oxedanda/avcad_final_project/main/data/processed/crop_productivity_clean.csv"

# Load datasets
df_labour = pd.read_csv(url_labour)
df_area = pd.read_csv(url_area)
df_crop = pd.read_csv(url_crop)

# Rename columns for easier use
df_area = df_area.rename(columns={"Agricultural holdings area (ha)": "Agricultural_Holdings_Area_ha"})
```

```python
display(df_labour.head())
display(df_area.head())
display(df_crop.head())
```

## 3.1 Evolution of Agricultural Labour in Mainland Portugal

### Agricultural Labour Data Table

Here is a table generated from the `agricultural_labour_clean.csv` data.

```python
import matplotlib.pyplot as plt

# Select relevant columns for the table
df_labour_table = df_labour[['year', 'Agricultural labour']].copy()

# Rename columns for proper display
df_labour_table = df_labour_table.rename(columns={
    'year': 'Year',
    'Agricultural labour': 'Agricultural Labour'
})

# Display the table in the notebook
display(df_labour_table)

# Create a figure and an axes object
fig, ax = plt.subplots(figsize=(8, 2.5))

# Hide the axes
ax.axis('off')
ax.axis('tight')

# Create the table using matplotlib
table = ax.table(
    cellText=df_labour_table.values,
    colLabels=df_labour_table.columns,
    loc='center',
    cellLoc='center'
)

# Adjust font size and layout for better readability
table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1.2, 1.2)

plt.title('Agricultural Labour in Mainland Portugal (1989–2019)', loc='left', fontsize=12, pad=20)

# Save the table as a PNG image
plt.savefig('agricultural_labour_table.png', bbox_inches='tight', dpi=300)
print("The table has been saved as 'agricultural_labour_table.png'")

plt.close()
```

### Evolution of Agricultural Labour Line Graph

Here is a line graph visualizing the evolution of agricultural labour, with the y-axis starting at 0.

```python
# Set the style for the plot

fig, ax = plt.subplots(figsize=(10, 6))

sns.lineplot(
    x='year',
    y='Agricultural labour',
    data=df_labour,
    marker='o',
    ax=ax
)

ax.set_title('Evolution of Agricultural Labour in Mainland Portugal (1989–2019)')
ax.set_xlabel('Year')
ax.set_ylabel('Agricultural Labour')
ax.set_xticks(df_labour['year'])
ax.grid(True, linestyle='--', alpha=0.7)
ax.set_ylim(bottom=0)
ax.ticklabel_format(style='plain', axis='y')

plt.tight_layout()
plt.savefig('agricultural_labour_line_chart.png', dpi=300, bbox_inches='tight')
print("The line chart has been saved as 'agricultural_labour_line_chart.png'")
plt.show()
```

## 3.2 Changes in Agricultural Area and Land Use

Contrary to the sharp decline observed in agricultural labour, total agricultural area in mainland Portugal remained relatively stable over the 30-year period. The table presents the evolution of the total agricultural holdings area between 1989 and 2019.

```python
import matplotlib.pyplot as plt

# Select relevant columns for the table
df_area_table = df_area[['year', 'Agricultural_Holdings_Area_ha']].copy()

# Rename columns for display
df_area_table = df_area_table.rename(columns={
    'year': 'Year',
    'Agricultural_Holdings_Area_ha': 'Agricultural Holdings Area (ha)'
})

# Display the table in the notebook
display(df_area_table)

# Create figure
fig, ax = plt.subplots(figsize=(8, 2))

# Hide axes
ax.axis('off')
ax.axis('tight')

# Create table
table = ax.table(
    cellText=df_area_table.values,
    colLabels=df_area_table.columns,
    loc='center',
    cellLoc='center'
)

# Format table
table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1.2, 1.2)

plt.title('Total Agricultural Holdings Area in Mainland Portugal (1989–2019)', loc='left', fontsize=12, pad=20)

# Save image
plt.savefig('agricultural_area_table.png', bbox_inches='tight', dpi=300)
print("The table has been saved as 'agricultural_area_table.png'")

plt.close()
```

#### Evolution of Total Agricultural Area

Here is a bar and a line chart visualizing the evolution of total agricultural area, with the y-axis starting at 0:




```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
from PIL import Image

sns.set_style('whitegrid')

# Bar chart
df_area_bar = pd.DataFrame({
    'year': [1989, 1999, 2009, 2019],
    'Agricultural_Holdings_Area_ha': [5157213, 5039569, 4571531, 4987658]
})

fig_bar, ax_bar = plt.subplots(figsize=(10, 6))

sns.barplot(
    x='year',
    y='Agricultural_Holdings_Area_ha',
    data=df_area_bar,
    color='gray',
    ax=ax_bar
)

ax_bar.set_title('Evolution of Total Agricultural Area in Mainland Portugal (1989–2019)')
ax_bar.set_xlabel('Year')
ax_bar.set_ylabel('Agricultural Holdings Area (ha)')
ax_bar.set_ylim(bottom=0)
ax_bar.ticklabel_format(style='plain', axis='y')
ax_bar.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.savefig('total_agricultural_area_bar_chart.png', dpi=300, bbox_inches='tight')
print("The bar chart has been saved as 'total_agricultural_area_bar_chart.png'")
plt.close(fig_bar)

# Line chart
fig_line, ax_line = plt.subplots(figsize=(10, 6))

sns.lineplot(
    x='year',
    y='Agricultural_Holdings_Area_ha',
    data=df_area,
    marker='o',
    ax=ax_line
)

ax_line.set_title('Evolution of Total Agricultural Area in Mainland Portugal (1989–2019)')
ax_line.set_xlabel('Year')
ax_line.set_ylabel('Agricultural Holdings Area (ha)')
ax_line.set_xticks(df_area['year'])
ax_line.grid(True, linestyle='--', alpha=0.7)
ax_line.set_ylim(bottom=0)
ax_line.ticklabel_format(style='plain', axis='y')

plt.tight_layout()
plt.savefig('total_agricultural_area_line_chart.png', dpi=300, bbox_inches='tight')
print("The line chart has been saved as 'total_agricultural_area_line_chart.png'")
plt.close(fig_line)

# Display saved images side by side
img_bar = Image.open('total_agricultural_area_bar_chart.png')
img_line = Image.open('total_agricultural_area_line_chart.png')

fig_display, axes = plt.subplots(1, 2, figsize=(16, 6))

axes[0].imshow(img_bar)
axes[0].axis('off')

axes[1].imshow(img_line)
axes[1].axis('off')

plt.tight_layout()
plt.show()
plt.close(fig_display)
```

#### 3.3 Land Use Concentration and Restructuring

```python
import pandas as pd
import matplotlib.pyplot as plt

data_avg_holding = {
    'Year': [1989, 1999, 2009, 2019],
    'Average Temp. Holding Size (ha)': [11.5, 18.0, 27.7, 39.9],
    'Average Perm. Holding Size (ha)': [11.4, 15.7, 20.2, 24.3]
}

df_avg_holding_size = pd.DataFrame(data_avg_holding)
df_avg_holding_size['Year'] = df_avg_holding_size['Year'].astype(int)

display(df_avg_holding_size)

fig, ax = plt.subplots(figsize=(8, 2.5))
ax.axis('off')
ax.axis('tight')

table = ax.table(
    cellText=df_avg_holding_size.values,
    colLabels=df_avg_holding_size.columns,
    loc='center',
    cellLoc='center'
)

table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1.2, 1.2)

plt.title('Average Holding Size by Crop Type (1989–2019)', loc='left', fontsize=12, pad=20)
plt.savefig('average_holding_size_table.png', bbox_inches='tight', dpi=300)
print("The table has been saved as 'average_holding_size_table.png'")

plt.close()
```

## Evolution of Average Holding Size by Crop Type (1989–2019)

Here is a line chart visualizing Evolution of Average Holding Size by Crop Type:






```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')

fig, ax = plt.subplots(figsize=(10, 6))

sns.lineplot(
    x='Year',
    y='Average Temp. Holding Size (ha)',
    data=df_avg_holding_size,
    marker='o',
    label='Average Temp. Holding Size (ha)',
    ax=ax
)

sns.lineplot(
    x='Year',
    y='Average Perm. Holding Size (ha)',
    data=df_avg_holding_size,
    marker='o',
    label='Average Perm. Holding Size (ha)',
    ax=ax
)

ax.set_title('Evolution of Average Holding Size by Crop Type (1989–2019)')
ax.set_xlabel('Year')
ax.set_ylabel('Average Holding Size (ha)')
ax.set_xticks(df_avg_holding_size['Year'])
ax.grid(True, linestyle='--', alpha=0.7)
ax.legend()
ax.set_ylim(bottom=0)

plt.tight_layout()
plt.savefig('average_holding_size_line_chart.png', dpi=300, bbox_inches='tight')
print("The line chart for average holding size has been saved as 'average_holding_size_line_chart.png'")
plt.show()
plt.close(fig)
```
